```python
import sagemaker
from sagemaker import get_execution_role
from sagemaker.sklearn.processing import ScriptProcessor
from sagemaker.workflow.steps import ProcessingStep
from sagemaker.workflow.pipeline import Pipeline
```

```python
role = get_execution_role()
sagemaker_session = sagemaker.Session()
default_bucket = sagemaker_session.default_bucket()

uri_script = f"s3://{default_bucket}/scripts/RPT/TEST_RPT_4A_validacion_tarde.py"

processor = ScriptProcessor(
    command=['python3'],
    image_uri=sagemaker.image_uris.retrieve("sklearn", sagemaker_session.boto_region_name, "1.2-1"),
    role=role,
    instance_count=1,
    instance_type="ml.m5.xlarge"
)

step_validacion_tarde = ProcessingStep(
    name="Validacion-Errores-Paises-Tarde",
    processor=processor,
    code=uri_script
)
```

```python
pipeline = Pipeline(
    name="Pipeline-Validacion-Errores-Paises-Tarde",
    steps=[step_validacion_tarde],
    sagemaker_session=sagemaker_session
)

pipeline.upsert(role_arn=role)
print("Pipeline registrado con éxito.")
```
